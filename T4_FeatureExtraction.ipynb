```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Memuat data yang telah di-preprocessing (hasil Tugas 3)
df = pd.read_csv('/content/drive/MyDrive/Muhammad Yahya Ayyasy_Tugas 21 Agustus 2026/data_preprocessed.csv')
```

```python
# 1. BANGUN REPRESENTASI BAG OF WORDS (BoW)
# ==========================================
bow = CountVectorizer(max_features=500)
# Menggunakan .fillna('') untuk mencegah error apabila ada baris yang kosong setelah pre-processing
X_bow = bow.fit_transform(df['teks_processed'].fillna(''))

print('Shape BoW:', X_bow.shape)

print('\n10 kata pertama di vocabulary:')
print(bow.get_feature_names_out()[:10])

# Konversi matriks BoW ke DataFrame pandas
df_bow = pd.DataFrame(X_bow.toarray(), columns=bow.get_feature_names_out())
print('\nMatriks BoW (5 baris pertama):')
print(df_bow.head())
```

```python
# 2. BANGUN REPRESENTASI TF-IDF
# ==========================================
tfidf = TfidfVectorizer(max_features=500)
X_tfidf = tfidf.fit_transform(df['teks_processed'].fillna(''))

print('\nShape TF-IDF:', X_tfidf.shape)

# Konversi matriks TF-IDF ke DataFrame pandas
df_tfidf = pd.DataFrame(X_tfidf.toarray(), columns=tfidf.get_feature_names_out())
print('\nMatriks TF-IDF (5 baris pertama, dibulatkan):')
print(df_tfidf.head().round(3))
```

```python
# 3. TAMPILKAN 10 KATA SKOR TF-IDF TERTINGGI
# ==========================================
mean_tfidf = df_tfidf.mean().sort_values(ascending=False)
print('\nTop 10 kata dengan TF-IDF rata-rata tertinggi:')
print(mean_tfidf.head(10))
```
