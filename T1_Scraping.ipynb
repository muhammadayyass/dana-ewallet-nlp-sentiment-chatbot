```python
!pip install google-play-scraper

from google_play_scraper import reviews, Sort
import pandas as pd

result, _ = reviews(
    'id.dana',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=11000
)

data = [{
    'id': r['reviewId'],
    'teks': r['content'],
    'tanggal': r['at'],
    'sumber': 'playstore'
} for r in result]

df = pd.DataFrame(data)

# TAMBAHAN WAJIB: cek dulu apakah benar sudah >=1000 SEBELUM lanjut (syarat jobsheet)
print('Total review berhasil ditarik:', len(df))
if len(df) < 1000:
    print('⚠️ PERINGATAN: belum mencapai 1000 baris, ulangi dengan count lebih besar atau ganti sort=Sort.NEWEST')
else:
    print('✅ Syarat minimal 1000 baris terpenuhi')

df.to_csv('scraped_playstore.csv', index=False)
df.head()
```
